# LISTS (CRUD & CORE OPERATIONS)

```python
nums = [4, 2, 8, 1, 5]
print("\n--- 2. LISTS (CRUD) ---")
print("Initial list:", nums)

# READ
print("First element:", nums[0])
print("Last element:", nums[-1])
print("Sliced list [1:4]:", nums[1:4])
```

```python
# APPEND / INSERT
nums.append(10)
print("After append(10):", nums)
nums.insert(1, 99)
print("After insert(1, 99):", nums)
```

```python
# DELETE
print(nums)
popped = nums.pop()
print("After pop() (removed value:", popped, "):", nums)
nums.remove(99)
print("After remove(99):", nums)
del nums[0]
print("After del nums[0]:", nums)
```

```python
# COPY
nums_copy = nums.copy()
print("Shallow copy:", nums_copy)
```

```python
# FILTER
filtered = [x for x in nums if x > 3]
print("Filtered list (x > 3):", filtered)
```

```python
# REARRANGE
nums.sort()
print("After sort():", nums)
nums.reverse()
print("After reverse():", nums)
```

# LIST AGGREGATIONS & STATISTICS (WITHOUT MODULES)

```python
data = [4, 2, 3, 1, 5, 5]
print("\n--- 3. STATISTICS (WITHOUT MODULES) ---")
print("Data:", data)

# BASIC AGGREGATIONS
total_count = len(data)
total_sum = sum(data)
average = total_sum / total_count
min_val = min(data)
max_val = max(data)

print("Count:", total_count)
print("Sum:", total_sum)
print("Average/Mean:", round(average, 2))
print("Minimum:", min_val)
print("Maximum:", max_val)
```

```python
# MEDIAN LOGIC
sorted_data = sorted(data)
mid = total_count // 2
if total_count % 2 == 0:
    median_val = (sorted_data[mid - 1] + sorted_data[mid]) / 2
else:
    median_val = sorted_data[mid]
print("Median:", median_val)
```

```python
# MODE LOGIC
counts = {}
for val in data:
    counts[val] = counts.get(val, 0) + 1
mode_val = max(counts, key=counts.get)
print("Mode:", mode_val)
```

```python
# VARIANCE & STANDARD DEVIATION LOGIC
mean_val = average
variance_val = sum((val - mean_val) ** 2 for val in data) / (total_count - 1)
stdev_val = variance_val ** 0.5
print("Variance:", round(variance_val, 2))
print("Standard Deviation:", round(stdev_val, 2))
```

# LIST AGGREGATIONS & STATISTICS (WITH BUILT-IN MODULE)

```python
import statistics

print("\n--- 4. STATISTICS (WITH MODULE) ---")
print("statistics.mean():", round(statistics.mean(data), 2))
print("statistics.median():", statistics.median(data))
print("statistics.mode():", statistics.mode(data))
print("statistics.variance():", round(statistics.variance(data), 2))
print("statistics.stdev():", round(statistics.stdev(data), 2))
```
